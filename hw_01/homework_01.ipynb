## Задание 1
Пользователь в бесконечном цикле вводит числа через консоль до тех пор, пока не введёт что-то, отличное от числа (после этого цикл завершается). В конце программа должна вывести сумму всех введённых чисел.

```python
summa = 0

while True:
    inp = input('Type a number: ')
    try:
       num = float(inp.replace(' ', '')) #чек, удовлетворяет ли инпут условию
       summa += num #считаю сумму чисел
    except ValueError:
        print(f'"Warning! Input must be a number')
        break #брейк если инпут не число

print(f'Sum is: {summa}')
```

## Задание 2
Пользователь вводит строку через консоль (состоящую только из букв в разном регистре). Программа должна создать новую строку, содержащую только те буквы из исходной строки, которые были в верхнем регистре.


```python
inp_str = input('Type a string of alphas: ')
up_str = ''

if inp_str.isalpha(): #чек, удовлетворяет ли инпут условию
    for char in inp_str:
        if char.isupper():
            up_str += char #добавляю символ в up_str, если это заглавная буква
    if up_str == '': #чек, пустая ли up_str. Если да - подсвечиваю
        print (f'Warning! No uppercase letters in string "{inp_str}"')
    print(up_str)
else:
    print('Warning! Not only alphas, try again')
```

## Задание 3
Классическая игра FizzBuzz. Программа в цикле выводит в консоль числа от 1 до 100.

* Если число кратно 3, вместо него выводится Fizz.
* Если кратно 5 — Buzz.
* Если кратно и 3, и 5 — FizzBuzz.

```python
for i in range(1, 101):
    if i % 3 == 0 and i % 5 == 0: #чек, чему кратен элемент в генераторе
        print('FizzBuzz')
    elif i % 3 == 0:
        print('Fizz')
    elif i % 5 == 0:
        print('Buzz')
    else: #не было в задании, чисто для информативности, чтобы не выводить пустую строку
        print('No fizz or buzz')
```

## Задание 4
Пользователь вводит строку, содержащую любые символы (буквы, цифры и специальные символы). Программа должна вычислить и вывести сумму всех цифр, присутствующих в этой строке.

```python
inp_str = input('Type a string: ')
summa = 0
has_digit = False

for char in inp_str:
    try:
        char = int(char) #чек, является ли элемент строки целым числом. Если да - прибавляю его в переменную
        summa += char
        has_digit = True
    except ValueError:
        pass

if has_digit: #информирую, если в строке вообще нет цифр
    print(f'Sum is: {summa}')
else:
   print('Warning! There are no nums in ur string, nothing to sum') 
```

## Задание 5
Пользователь вводит через консоль 10 чисел. Программа должна определить максимальное число и вывести его в консоль.

```python
nums = input('Type 10 numbers sep by space: ').split()
isnumber = True
max_num = float('-inf')

if len(nums) != 10:
    print('Warning! Type exactly 10 numbers')
else:
    for num in nums:
        try: 
            num = float(num) #чек, является ли элемент списка числом
            if num > max_num:
                max_num = num #перебором определяю максимальное число
        except ValueError:
            print('Warning! Not only numbers')
            isnumber = False
            break #если элемент не число - прерываю цикл
    
    if isnumber:
        print(f'Max value: {max_num}')
```

## Задание 1: Зеркальное отражение цифр
Пользователь вводит пятизначное число. Программа должна зеркально отразить три центральные цифры. Первая и последняя остаются на местах.

Примеры:
* 23456 -> 25436
* 30789 -> 38709

```python
inp_num = input('Type 5-digit num: ')

if len(inp_num) == 5 and inp_num.isdigit(): #чек, удовлетворяет ли инпут условию ввода
    mid = inp_num[1:4] #через срез выделяю цифры посередине. Их буду отражать через отрицателбный шаг
    rev_num = inp_num[0] + mid[::-1] + inp_num[4]
    rev_num = int(rev_num)
    print(f'Reversed num is {rev_num}')
else:
    print('Warning! Type only 5-digit num:')
```

## Задание 2: Подсчет выходных до отпуска
Пользователь вводит сколько дней осталось до ближайшего отпуска. Программа должна вывести количество выходных дней до отпуска, если учесть, что выходные это суббота и воскресенье, сегодня понедельник и праздники мы не учитываем.

Примеры:
* 4  -> 0
* 6  -> 1
* 14 -> 4

```python
# Вариант 1 - с использованием библиотеки datetime

import datetime as dt

weekend_count = 0
try:
    days = int(input('Days till vacation: '))
    if days >= 0:
        current = dt.date(2026, 2, 16) #была идея через while посчитать ближайший пн .weekday()==0 от текущего дня .today()
        delta = current + dt.timedelta(days = days) #считаю дельту: сколько дней до 1го дня отпуска от текущего дня 
        while current < delta: #запускаю цикл от текущего дня до 1го дня отпуска не включительно
            if current.weekday() == 5 or current.weekday() == 6: #если значение текущего дня сб-5 или вс-6 - прибавляю к количеству выходных 
                weekend_count +=1
            current += dt.timedelta(days = 1)
        print(f'{weekend_count} weekends till vacation')
    else:
        print('Warning! Digit must be >= 0')
except ValueError:
    print('Warning! Type only int num')
```

```python
# Вариант 2 - без импорта библиотеки

weekend_count = 0

try:
    days = int(input('Days till vacation: '))
    if days >= 0:
        for day in range(days):
            dayNum = day % 7 #выделяю каждую новую неделю через нумерацию дней недели по остатку от деления:  от пн-0 до вс-6
            if dayNum == 5 or dayNum == 6:# если день равен 5(сб) или 6(вс) - прибавляю к количеству выходных
                weekend_count += 1
        print(f'{weekend_count} weekends till vacation')
    else:
        print('Warning! Digit must be >= 0')
except ValueError:
    print('Warning! Type only int num')
```

## Задание 3: Плитка шоколада
Пользователь вводит длину и ширину плитки шоколада, а также размер куска, который хочет отломить. Программа должна вычислить - можно ли совершить подобный разлом или нет, если учесть, что ломать можно только по прямой.

Примеры:
* 3, 4, 6  -> True
* 5, 7, 8  -> False
* 4, 5, 12 -> True

```python
can_take = False

try:
    length, width, size = int(input('Chokolate length: ')), int(input('Chokolate width: ')), int(input('Size to eat: ')) #чек, являются ли импуты типом инт
    if 0 < size < length*width: #чек, адекватны ли инпуты - числа больше 0 и размер кусочка < размера шоколадки
        if (size % length == 0 and size // length < width) or (size % width == 0 and size // width < length): #чек, можно ли отломить плитку целиком по длине или ширине
            can_take = True
        print(can_take)
    else:
        print("Warning! Wrong parameters")
except ValueError:
    print("Warning! Type only int num")
```

## Задание 4: Римские числа
Пользователь вводит целое положительное число, программа должна вернуть строку в виде римского числа.

Примеры:
* 3   -> III
* 15  -> XV
* 234 -> CCXXXIV

```python
rome_lst = [[1000, 'M'], [900, 'CM'], [500 , 'D'], [400, 'CD'], [100, 'C'], [90, 'XC'],  [50, 'L'], [40, 'XL'], [10, 'X'], [9, 'IX'], [5, 'V'], [4, 'IV'], [1, 'I']]
rome = ''

try:
    arab = int(input('Type int > 0: '))
    if arab <= 0:
        print('Warning! Num must be > 0')
    else:
        for val, symbol in rome_lst:
            while arab >= val:
                rome += symbol
                arab -= val
        print(f"Результат: {rome}")
except ValueError:
    print('Warning! Num must be int')
```

## Задание 5: Проверка на вещественное число
Пользователь вводит данные, проверить - являются ли они положительным вещественным числом. Не использовать встроенные функции для проверки, только методы данных и конструкцию IF.

Примеры:
* 5.6  -> True
* .78  -> True
* .67. -> False
* 5    -> True

Дополнительное задание, по желанию - проверка на отрицательные вещественные числа

```python
def is_float(inp_str): #возвращает проверку вещественности инпута
    isfloat = False

    if inp_str == '' or inp_str.count('.') > 1: #если инпут - пустая строка или точек больше одной - это не вещ число
        return isfloat
    
    rest_str = inp_str[1:] if inp_str.startswith('-') else inp_str #отсекаю "-" в начале, при его наличии

    if rest_str == '' or rest_str == '.' or rest_str.isspace(): #чек, не остался ли остаток строки пустым или только с точкой/пробелами после отсечения "-"
        return isfloat
    
    if rest_str.count('.') == 1: #если в остатке одна точка - разделяю его на две части
        left_str, right_str = rest_str.split('.')
        isfloat = (left_str.isdigit() or left_str == '') and right_str.isdigit() #если все правда - изфолат=тру
    elif rest_str.count('.') == 0:
        isfloat = rest_str.isdigit() #если все правда - изфолат=тру
    return isfloat

def is_negative(inp_str): #возвращает проверку отрицательности вещественного числа
    return inp_str.startswith('-') and inp_str != '-'
    
num = input('Type a num: ')

is_float_num = is_float(num)
is_neg_num = is_negative(num)

print('Is number float: ', is_float_num)
if is_float_num:
    print('Is number negative: ', is_neg_num)
```
