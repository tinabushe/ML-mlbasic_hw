**Шаг 1.** <br>

Прямая задается линейным уравнением $y = ax + b$. Т.к. моя прямая пороходит через начало координат (0, 0), то b=0, следавотельно уравнение можно упростить до $y= ax$, где

* $y$ - целевая переменнагя

* $a$ - искомый угловой коэффициент, задающий наклон прямой. Т.о. надо найти такой угловой коэфициент, при котором отклонение от точек будет минимальное

**Шаг 2.**<br>
Запишите целевую функцию, зависящую от этих параметров.

По заданию целевая фукнция считает суммарную ошибку/квадрат ошибки, т.е. сумму разниц между реальными значениями $y_i$ и предсказаниями $\hat{y}_i$. В моем случае $\hat{y}_i = a \cdot x_i$. Тогда ошибка для каждой конкретной точки будет: $(y_i - ax_i)$

Итого, целевая функция будет: 
* Сумма отклонений от данных точек по `у`: $\sum_{i=1}^{n} |y_i - ax_i|$: 
    * $|2 - a \cdot (-3)| = |2 + 3a|$
    * $|2 - 2a|$
    * $|-3 - 5a|$
    
    => $g(x) = |y_i - ax_i| = |2 + 3a| + |2 - 2a| + |-3 - 5a|$
* Сумма квадратов отклонений от данных точек по `у`: $\sum_{i=1}^{n} (y_i - ax_i)^2$
    * $(2 - a \cdot (-3))^2 = (2 + 3a)^2$
    * $(2 - 2a)^2$
    * $(-3 - 5a)^2$
    
    => $f(x) = (2 + 3a)^2 + (2 - 2a)^2 + (-3 - 5a)^2 = 4 + 12a + 9a^2 + 4 - 8a + 4a^2 + 9 + 30a + 25a^2 = 38a^2 + 34a + 17$

Самопроверка:

```python
from fractions import Fraction

class GaussJordan:
    def __init__(self, row_matrix: list[list[float]]) -> None:
        '''
        Инициализирует объект класса GaussJordan и преобразует элементы в точные дроби.

        Принимает двумерный список чисел и преобоазует каждый элемент 
        к точной дроби - fractions.Fraction, чтобы исключить накопление погрешностей 
        округления при использованеи остальных методов класса.
        '''
        self.matrix = [[Fraction(el) for el in row] for row in row_matrix]

    def adding(self, Rsrs_idx, Rdest_idx, k: int | float | Fraction = 1) -> None:
        '''
        Прибавляет к одной строке матрицы другую строкуэтой же матрицы, умноженную на коэффициент.
        Индексация строк начинается с 0.

        Args:
            Rsrs_idx: Индекс изменяемой строки (целое число от 0).
            Rdest_idx: Индекс строки-донора (целое число от 0).
            k: Коэффициент умножения. По умолчанию равен 1. Может быть передан как int-float-Fraction, но внутри метода 
                автоматически приводится к Fraction во имя точности вычислений.
        '''
        k = Fraction(k)
        for idx in range(len(self.matrix[Rsrs_idx])):
            self.matrix[Rsrs_idx][idx] += k * self.matrix[Rdest_idx][idx]

    def scalling(self, R_idx, k: int | float | Fraction = 1) -> None:
        '''
        Умножает все элементы выбранной строки матрицы на коэффициент.

        Используется для получения ведущей единицы в строке или для изменения знаков элементов строки.
        Индексация строк начинается с 0.

        Args:
            R_idx: Индекс строки матрицы, которую нужно отмасштабировать.
            k: Коэффициент умножения. По умолчанию равен 1. Может быть передан как int-float-Fraction, но внутри метода 
                автоматически приводится к Fraction во имя точности вычислений.
        '''
        k = Fraction(k)
        for idx in range(len(self.matrix[R_idx])):
            self.matrix[R_idx][idx] *= k
    
    def show(self):
        '''
        Кастомный метод вывода.
        
        Отображает матрицу в более-менее читабельном виде.
        '''
        print('\n'.join(str(row) for row in self.matrix))
```

```python
# исходная матрица A в виде двумерного спискв
raw_data = [
    [1, 2, 4, 1, 0, 0],
    [2, 1, -1, 0, 1, 0],
    [1, -1, 1, 0, 0, 1]
]

matrix_AE = GaussJordan(raw_data)
matrix_AE.show()
```
