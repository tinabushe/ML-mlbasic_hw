### В модуле `plane` создайте класс `Plane`
* класс `Plane` должен быть наследником `Vehicle`
* добавьте атрибуты `cargo` и `max_cargo` классу `Plane`
* добавьте `max_cargo` в инициализатор (переопределите родительский)
* объявите метод `load_cargo`, который принимает число, проверяет, что в сумме с текущим `cargo` не будет перегруза, и обновляет значение, в ином случае выкидывает исключение `exceptions.CargoOverload`
* объявите метод `remove_all_cargo`, который обнуляет значение `cargo` и возвращает значение `cargo`, которое было до обнуления

````python
from hw_05.base import Vehicle
from hw_05 import exceptions as ex

class Plane(Vehicle):

    def __init__(self, weight, fuel, fuel_consumption, max_cargo):
        super().__init__(weight, fuel, fuel_consumption)
        self.cargo = 0
        self.max_cargo = max_cargo

    def load_cargo(self, load):
        if self.cargo + load > self.max_cargo:
            raise ex.CargoOverload()
        self.cargo += load

    def remove_all_cargo(self):
        cargo_before_removal = self.cargo
        self.cargo = 0
        return cargo_before_removal
````

##### <span style="color:#49A7E5">как это должно работать, как я поняла:</span>

У меня есть милион атрибутов и параметров связанных друг с другом, родителями и объектами. Создавать сущность в том или ином месте надо опираясь на такую логику:

**1. Когда создавать атрибут класса**<br>
если значение ВСЕГДА ОБЩЕЕ для всех экземпляров, напр:
* у ВСЕХ машин 4 колеса
* у ВСЕХ самолетов 2 крыла

**2. Когда задавать параметр в инициализаторе (в скобках)?**<br>
если значение УНИКАЛЬНО для каждого объекта и НЕИЗВЕСТНО заранее, напр: 
* одна машина - волга весом тонну, другая машина - мерс, весом 2 тонны
* один самолет - кукурузник весом 3 тонны, другой самолет - боинг весом 10 тонн

**3. Когда жестко задавать значение внутри метода, не передавая в скобки?**<br>
если ДЛЯ ВСЕХ значение одинаковое НА СТАРТЕ, но оно будет В БУДУЩЕМ МЕНЯТЬСЯ, напр:
* любой самолет на старте будет пустой, потом - каждый будут загружать по разному => self.cargo = 0 ---> self.cargo = N

**4. Когда задавать параметр только внутри метода (не "_инит"_)?**<br>
если данные ВРЕМЕННЫЕ, нужны В МОМЕНТЕ, напр:
* в какой-то самолет загрузили N груза. N - число нужное в моменте, для рассчета потенциального перегруза => load нужен внутри "load_cargo(self, load):" для проврки условия " > self.max_cargo"

````python
from hw_05.car import Car
from hw_05.engine import Engine
from hw_05 import exceptions as ex

def test_drive_positive():
    # создала машину: вес 1000, топливо 100, расход 10
    my_car = Car(1000, 100, 10)
    print(f'1. Cоздала машину. Толплвиа на старте:  {my_car.fuel} л')

    bad_car_eng = Engine(2.5, 6)
    good_car_eng = Engine(10.5, 20)

    try:
        my_car.set_engine(bad_car_eng)
        print(f'\nПлохой движок с параметрами: {bad_car_eng.volume}, {bad_car_eng.pistons}')

        my_car.set_engine(good_car_eng)
        print(f'Хороший движок с параметрами: {good_car_eng.volume}, {good_car_eng.pistons}\n')

        # зарвожу двигатель
        my_car.start()
        print('2. Завеласть, self.started = True')
        
        # хочу проехать 5 км (расстояние * на расход = 7 * 10 => надо 70 топлива)
        distance = 7
        # топлива на поездку: топливо - топливо в полете = 100 - 70 = 30 => остаток 30л
        my_car.move(distance)
        print(f'3. Прилетела. Остаток топлива: {my_car.fuel} л')
    
    except (ex.LowFuelError, ex.NotEnoughFuel, ex.CargoOverload) as e:
        print(f'Ошибка: {e}')

if __name__ == '__main__':
    test_drive_positive()
````

````python
from hw_05.plane import Plane
from hw_05 import exceptions as ex

def test_flight_positive():
    # создала самолет: вес 10000, топливо 3000, расход 50, макс. груз 2000
    my_plane = Plane(10000, 3000, 50, 2000)
    print(f'1. Cоздала самолет, не грузилась. Толплвиа на старте:  {my_plane.fuel} л, вес груза: {my_plane.cargo} кг')

    try:

        # хагружаю самолет на 1500 кг груза
        my_plane.load_cargo(1500)
        print(f'2. Загрузилась. Вес груза: {my_plane.cargo} кг')

        # зарвожу двигатель
        my_plane.start()
        print('3. Завеласть, self.started = True')

        # хочу пролететь 50 км (расстояние * на расход = 50 * 50 => надо 2500 топлива)
        distance = 50
        # топлива на полет: топливо - топливо в полете = 3000 - 2500 = 500 => остаток 500л
        my_plane.move(distance)
        print(f'4. Прилетела. Остаток топлива: {my_plane.fuel} л')

        # разгружаюсь по прибытии
        removed_cargo = my_plane.remove_all_cargo()
        print(f'5. Разгрузилась. Выгружено: {removed_cargo} кг. На борту осталось груза на: {my_plane.cargo} кг')

    except (ex.LowFuelError, ex.NotEnoughFuel, ex.CargoOverload) as e:
        print(f'Ошибка: {e}')

if __name__ == '__main__':
    test_flight_positive()
````
