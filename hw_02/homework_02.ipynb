## Задание 3
Написать программу, которая удаляет из списка все дубликаты, сохранив исходный порядок элементов.


````python
import random

rand_lst = [random.randint(1,7) for _ in range(10)] #генерация рандомного списка
seen = set()
uniq_lst = []
for num in rand_lst: #последовательно перебираю элементы оригинального списка
    if num not in seen: #если элемента нет во множестве - добавляю и в сет и в юник_лст
        seen.add(num)
        uniq_lst.append(num)

print(rand_lst)
print(uniq_lst)
````

## Задание 1: Получение однозначного числа
Пользователь вводит целое число, программа складывает все цифры числа, с полученным числом — то же самое, и так до тех пор, пока не получится однозначное число.

Примеры:
* 545   -> 5
* 12345 -> 6

````python
def elements_sum(any_lst): #считаю и возвращаю сумму элементов списка
    summa = 0
    for i in any_lst:
        summa += i
    return summa

inp_str = input('Type a num >= 0: ')

if inp_str.isdigit(): #чек, что в инпуте только цифры
    int_lst = [int(num) for num in inp_str]
    int_sum = elements_sum(int_lst) # считаю сумму элементов в функцию и возвращаю в переменную, хранящую сумму
    while int_sum > 9: #пока сумма > 9, в цикле собирается список из цифр в int_sum, обернутой в строку, 
        int_lst = [int(digit) for digit in str(int_sum)]
        int_sum = elements_sum(int_lst) #в функцию отправляется интовый список и возвращается сумма его элементов
        
else:
    print('Warning! Type digit num')

print(int_sum)
````


## Задание 2: Кинотеатр
Дан список списков, каждый вложенный список состоит из 1 и 0, количество вложенных списков — количество рядов. Пользователь вводит, сколько билетов ему требуется. Программа должна найти ряд, где можно приобрести нужное количество билетов (места должны быть рядом). Если таких рядов несколько, то ближайший к экрану (ближайшим считается нулевой ряд). Если таких мест нет, то вывести False.

Примеры:
* [[0,1,1,0], [1,0,0,0], [0,1,0,0]], 2 -> 1
* [[0,1,1,0], [1,0,1,0], [1,1,0,1]], 2 -> False

````python
rows_lst = [[0,1,1,0], [1,1,0,0], [0,0,0,1]]

tickets = 3
isSpace = False

for irow, row in enumerate(rows_lst): #ищу первый подходящий ряд
    count = 0
    for seat in row: #пытаюсь найти нужное кол-во мест в ряду
        if seat == 0:
            count += 1
            if count == tickets: #когда колво свободных сидений "0" будет равно кол-ву тикетов - цикл прерывается
                isSpace = True
                break
        else:
            count = 0
    if isSpace:
        break

if isSpace: #если места найдены - возвращается номер ряда, на котором расположены места. Иначе - False
    print(irow)
else:
    print(isSpace)
````


## Задание 3: Алгоритм RLE
Необходимо написать упрощенную версию алгоритма RLE. Алгоритм RLE объединяет подряд идущие символы в коэффициент и символ.

Примеры:
* asssdddsssddd -> 1a3s3d3s3d
* aaabbbbccccc  -> 3a4b5c

````python
inp_str = input('Type smth: ') #не привожу к одному регистру, чтобы строчные и заглавные буквы считались отдельно
new_str = ''
count_char = 1 #отсчет индексов с единицы, чтобы не потерялся первый элемент при сравнении "текущий-предыдущий"
if inp_str == '':
    print('Warning! String cant be empty')
else:
    for char in range(1, len(inp_str)):
        if inp_str[char] == inp_str[char - 1]: #сравниваю текущий и предыдущий элементы, если равны - увеличиваю счетчик
            count_char +=1
        else: #если не равны - записываю в новую строку накопленный счетчик и предыдущий элемент. Для нового сравнения "текущий-предыдущий" сбрасываю счетчик
            new_str += str(count_char) + inp_str[char-1]
            count_char = 1

    new_str += str(count_char) + inp_str[-1] #записываю в новую строку последний элемент в инпуте, тк он не попадает в сравнение как "предыдущий"
    print(new_str)
````


## Задание 4: Шифр Цезаря
Реализуйте программу для шифрования текста с помощью шифра Цезаря.
Шифр Цезаря — это метод шифрования, при котором каждая буква в тексте заменяется буквой, стоящей на фиксированное число позиций дальше в алфавите.

**Требования**
1. Программа должна принимать на вход:

* Строку текста для шифрования
* Ключ шифра (число от 0 до 25)
* Правила шифрования:

2. Шифруются только латинские буквы (a-z, A-Z)
* Сохраняется регистр букв (заглавные остаются заглавными)
* Пробелы, цифры и знаки препинания остаются без изменений
* Сдвиг циклический (после 'z' идет 'a', после 'Z' идет 'A')

**Алгоритм**

Для каждого символа в строке:
* Если это латинская буква, найти её позицию в алфавите
* Добавить к позиции значение ключа
* Если результат больше 25, использовать остаток от деления на 26
* Преобразовать обратно в букву, сохранив регистр
* Если это не буква, оставить символ без изменений

Примеры:
* "Hello World", 3      -> "Khoor Zruog"
* "Python is great!", 5 -> "Udymts nx lwjfy!"
* "XYZ abc", 3          -> "ABC def"

````python
import string

def newIdx(key, idx, alpha_dict): #ищет позицию смещения буквы из инпута в списке с алфавитом на размер ключа
    new_idx = (idx + key) % len(alpha_dict) #если смещение буквы на размер ключа превышает длину алфавита, остатком от деления "обнуляю" ее, индекс продолжает считается с начала алфавита
    return new_idx

latin_lower_alphas = list(string.ascii_lowercase)
latin_upper_alphas = list(string.ascii_uppercase)
inp_str, new_str = input('Type smth: '), ''

try:
    key = int(input('Type a 0 <= key <= 25: ')) #чек, что ключ в заданных границах
    if key in range(0, 26):
        for char in inp_str:
            if char.isalpha() and char in latin_lower_alphas: #чекаю, является ли элемент буквой и входит ли он в алфавит СТРОЧНЫХ букв алфавита, если нет - либо в ЗАГЛАВНЫХ, либо НЕАЛФАВИТНЫЙ символ
                latin_char_idx = latin_lower_alphas.index(char)  #ищу номер буквы из инпута в алфавите, он уйдет в функцию для поиска позиции смещения. Когда нашла позицию смещения - 
                char = latin_lower_alphas[newIdx(key, latin_char_idx, latin_lower_alphas)] #- присваиваю в переменную буквы из инпут_строки букву, которая в алфавите находится на позиции смещения, добавляю переменную в новую строку
                new_str += char
            elif char.isalpha() and char in latin_upper_alphas: #чекаю, является ли элемент буквой и входит ли он в алфавит ЗАГЛАВНЫХ букв алфавита, если нет - НЕАЛФАВИТНЫЙ символ
                latin_char_idx = latin_upper_alphas.index(char)
                char = latin_upper_alphas[newIdx(key, latin_char_idx, latin_upper_alphas)]
                new_str += char
            else: # методм исключения НЕАЛФАВИТНЫЙ символ, добавляю его как есть
                new_str += char
        print('full str: ', new_str)
    else:
        print('Warning! Key must be in [0, 25]')
except ValueError:
    print('Warning! Key must be int')
````

````python
diary = {}

while True:
    inp_str = input('Type "название предмета фамилия ученика оценка" or press Enter to stop the loop: ')
    if inp_str != '': #для прерывания цикла
        allinfo = inp_str.split() #записываю инпут в список
        if len(allinfo) == 3: #если введено 3 значения - дроблю список на строки и проверяю входимость оценки 
            lesson, last_name, score = allinfo
            if score.isdigit() and int(score) in range(1,6):
                    lesson, last_name = lesson.capitalize(), last_name.capitalize()
                    score = int(score)
                    if lesson not in diary: # если предмет или фамилия ученика уже есть в словаре - сразу добавляю оценку для ученика, если еще нет сначала создаю предмет/фамилию и дбавляю оценку
                        diary[lesson] = {}
                    if last_name not in diary[lesson]:
                        diary[lesson][last_name] = []
                    diary[lesson][last_name].append(score) 
            else:
                print('Warning! Score must be inp from 1 to 5, try again')
        else:
            print('Warning! Wrong input, try again')
    else:
        if diary == {}: #если словарь не заполнялся - информирую, что он пустой
            print('Diary is empty')
        else: #если словарь заполнялся - один из вариантов отображения данных
            # Вариант 1 - по тексту задания
            # for lesson in diary.keys():
            #     print(f'\n{lesson}')
            #     for last_name, score in diary[lesson].items():
            #         score_str = ' '.join(map(str, score))
            #         print(f'{last_name:10} {score_str}')
            # Вариант 2 - по тексту примера
            checkClass = input('You stopped the loop.\nType class name: ').capitalize()
            if checkClass in diary:
                print(f'\n{checkClass}')
                for last_name, score in diary[checkClass].items():
                    score_str = ' '.join(map(str, score))
                    print(f'{last_name:10} {score_str}')
            else:
                print('Warning! No such class name')
        break
````
