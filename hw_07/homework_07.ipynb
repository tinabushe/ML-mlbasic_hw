### Создайте одномерный массив размера 10, заполненный нулями и пятым элемент равным 1. Трансформируйте в двумерный массив.

```python
data_arrow = np.array([1 if i % 5 == 0 else 0 for i in range(1,11)])
print('Одномерный массив: ', data_arrow)

print('\n--Двумерный массив--')
data_arrow1 = data_arrow.reshape(2, 5)
print('Двумерный массив 2х5:\n', data_arrow1)
data_arrow2 = data_arrow.reshape(5, 2)
print('Двумерный массив 5х2:\n', data_arrow2) # --- refactor: data_arrow1 -> data_arrow2 ---
```

### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
data_arrow = np.arange(10,50)
print('Одномерный массив:\n', data_arrow)

data_arrow_rev = data_arrow[::-1]
print('Одномерный массив перевернутый:\n', data_arrow_rev)

print('\n--Только четные числа:--')
data_arrow_rev_even1 = data_arrow_rev[1::2]
print('Одномерный массив перевернутый, срез:\n', data_arrow_rev_even1)
data_arrow_rev_even2 = data_arrow_rev[data_arrow_rev % 2 ==0]
print('Одномерный массив перевернутый, остаток от деления:\n', data_arrow_rev_even2) # --- refactor: data_arrow_rev_even1 -> data_arrow_rev_even2 ---
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
from random import randint

data_arrow_ran = np.array([randint(1, 50) for _ in range(24)]).reshape(4, 3, 2)
print('Массив 4х3х2:\n', data_arrow_ran)
print('Минимум: ', data_arrow_ran.min())
print('Максимум: ', data_arrow_ran.max())
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
arrow1 = np.arange(0,24).reshape(6, 4)
print('Массив 6х4:\n', arrow1)

arrow2 = np.arange(24,36).reshape(4, 3)
print('Массив 4х3:\n', arrow2)

print('Матричное умножение:\n', arrow1@arrow2)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
ran_arrow = np.array([randint(1, 50) for _ in range(49)]).reshape(7, 7)
print('Массив 7х7:\n', ran_arrow)

print('\n--Вычисления--')
mean_el = ran_arrow.mean().round(2)
std_el = ran_arrow.std().round(2)
print('Среднее арифметическое:', mean_el)
print('Среднеквадратичное отклонение:', std_el)

z_arrow_ran = (ran_arrow - mean_el) / std_el
print('Стандартизированный массив 7х7:\n', z_arrow_ran)
```
