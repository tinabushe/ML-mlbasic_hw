## Задание 1
Написать функцию, которая принимает неограниченное количество чисел в виде позиционных аргументов и ключевой аргумент — операцию над этими числами (сложение или умножение).
Функция должна возвращать результат выполнения указанной операции.

```python
def plus_or_mult_args(*args, operation='plus'):#выполняет арифметическую операцию сложения или умножения элементов, переданных в *аргс, и возвращает ее результат
    if not args:
        return 'Warning! Empty input' #информирую,  если аргументы не были переданы
    
    if operation == 'plus': # если была передана операция плас, элементы складываются, если мульт - перемножаются
        res = 0
        for num in args:
            res += num
        return res
    elif operation == 'mult':
        res = 1
        for num in args:
            res *= num
        return res
    else:
        return 'Warning! No such operation type'


str_lst = input('Type nums separated by spase: ').split()
int_operation = input('Type operation type - plus OR mult: ').lower()

try:
    numLst = [float(char) for char in str_lst] #чек, что элементы инпута счиатются числовыми. Если да - передаю элементы и ключ операции в функцию
    result = plus_or_mult_args(*numLst, operation = int_operation)
    print(result)
except ValueError:
    print('Warning! Wrong input')
```

## Задание 3
Написать функцию, которая создаёт абсолютный путь к файлу.

Позиционные аргументы:

* название диска,
* неограниченное количество папок,
* имя файла (без расширения).


Ключевые аргументы:
* ext — расширение файла,
* sep — разделитель (по умолчанию '/').


Пример:
* full_path('c:', 'work', 'python', 'function', 'main', ext='py') ➜ 'c:/work/python/function/main.py'

```python
def absolute_path(diskname, *path_parts, ext = '', sep = '/'): #возвращает абсолютный путь к файлу
    if ':' not in diskname:
        diskname += ':'
    path = diskname + sep #задаю путь: диск+разделитель
    
    if not path_parts: #если папки отсуствуют - возвращается только корневая папка
        return path
    
    *folds, filename = path_parts #папки = все элементы кортежа, кроме последнего. Имя файла = последний элемент

    folds = [f for f in folds if f] # фильтрация пустых строк: if f - false если f==''
    
    if folds:
        folds = sep.join(folds) #если есть папки - форматирую папки к виду абсолютного пути
        path += folds + sep  #дополняю путь: папки+разделитель

    path += filename #дополняю путь: имя файла
    if ext and not filename.endswith('.' + ext):
        path += '.' + ext

    return path

abs_path = absolute_path('c', 'work', 'python', 'function', 'main', ext='py')
print(abs_path)
```

Написать функцию, которая принимает список, состоящий из объектов разных типов, и возвращает словарь, где:
* ключи — типы данных объектов;
* значения — списки объектов соответствующего типа.

```python
def type_define1(any_lst):
    type_dict = {}
    for el in any_lst:
        el_type = type(el) #тип элемента считается ключом
        if el_type not in type_dict.keys(): #если такого типа нет среди ключей - сначала он создается
            type_dict[el_type] = []
        type_dict[el_type].append(el) #в список значений ключа добалвяется элемент
    return type_dict


diff_type_lst = [1, True, [2, 'a'], 'hello', {1: 2}, {3, 4}, ('my', 1, 'tuple'), 2, False, [3, 'b'], 'world', {3: 4}, {5, 6}, ('my', 2, 'tuple')]
output_type_dict = type_define1(diff_type_lst)
print(output_type_dict)
```

## Задание 1: Конвертер регистров
Написать функцию, которая будет переводить snake_case в PascalCase и наоборот.

Функция должна сама определять - какой формат ей передали. Можно добавить ключевой аргумент, который будет принудительно возвращать один из форматов.

Примеры:

* otus_course     -> OtusCourse
* PythonIsTheBest -> python_is_the_best

```python
def formating(inp_str, inp_format = ''): #принимает инпут строку и ключ формата (опц) и возвращает аутпут строку в формате snake_case или PascalCase
    out_str = '' 
    issnake = '_' in inp_str #переменная для проверки на snake_case в инпут строке
    ispask = inp_str and inp_str[0].isupper() and '_' not in inp_str #переменная для проверки на PascalCase
    inp_str = inp_str.strip()

    if inp_format == 'psk' or (issnake and inp_format == ''): #если ключ формата = "пск", либо он пустой И в инпуте есть "_" - формат инпута будет snake_case
        words_lst = inp_str.split('_') #формирую список из элементов инпута с разделением по "_"
        out_str = ''.join(word.capitalize() for word in words_lst)
        return out_str
    
    elif inp_format == 'snk' or (ispask and inp_format == ''): #если ключ формата = "снк", либо он пустой И для инпута срабатывает проверка из ispask - формат инпута будет PascalCase
        for i, symb in enumerate(inp_str):
            if i == 0: #перевод первого элемента инпута к нижнему регистру
                out_str += symb.lower()
            else:
                if symb.isupper(): #если элемент инпута заглавная буква - привожу к нижнему регистру, добавляю подчеркивание и букву в аутпут строку
                    out_str += '_' + symb.lower()
                else:
                    out_str += symb #добавляю элемент в аутпут строку
        return out_str 
    else:
        return 'Warning! Input string must be snake_case or PascalCase format'

inp_str, inp_format = input('Type snake_case or PascalCase string: '), input('Type format to which to convert the string - snk or psk - or press "Enter" key for a forced convertion')
inp_format = inp_format.strip().lower()

if inp_format not in ['snk', 'psk', '']:
    print('Warning! Convertion format must be "snk", "psk". Or press "Enter" key for a forced convertion')
else:
    print(inp_str, '-->', formating(inp_str, inp_format))
```

## Задание 2: Проверка валидности даты
Написать функцию проверяющую валидность введенной даты.

Примеры:

* 29.02.2000 -> True
* 29.02.2001 -> False
* 31.04.1962 -> False

```python
from datetime import datetime

def isDateValid1(str_date):
    isvalid = False
    try:
        datetime.strptime(str_date, "%d.%m.%Y") #проверяется сразу все - и корректность формата ввода в инпут, и входимость чисел в диапазоны кол-ва дней в месяце и кол-ва месяцев в году, и високосность и т.д.
        isvalid = True #если валидация прошла успешно - инпут дата считается корректной, возвращается Тру
    except ValueError:
        pass
    return isvalid

inp_str = input('Type date in format DD.MM.YYYY')
print(isDateValid1(inp_str))
```

## Задание 3: Проверка на простое число
Функция проверки на простое число. Простые числа – это такие числа, которые делятся на себя и на единицу.


Примеры:
* 17 -> True
* 20 -> False
* 23 -> True

```python
from math import sqrt

def is_prime_simple(num):
    if num < 2: #числа < 2 не простые
        return False
    if num == 2 or num == 3:
        return True
    if num % 2 == 0:  # четные числа > 2 СОСТАВНЫЕ
        return False

    # проверка нечетных делителей: от 3 до корня из num с шагом "2" для учета всех нечетных
    for i in range(3, int(sqrt(num)) + 1, 2):
        if num % i == 0:
            return False  # найден делитель - СОСТАВНОЕ

    return True  # делителей нет - ПРОСТОЕ
```

## Задание 4: Учет пользователей
Пользователь в бесконечном цикле вводит данные пользователей: имя, затем фамилию, возраст и ID. Ввод продолжается до тех пор, пока не будет введено пустое поле.

Пользователи заносятся в словарь, где ключ это ID пользователя, а остальные данные записываются в виде кортежа.

Программа должна проверять:
* имя и фамилия состоят только из символов и начинаются с большой буквы - если не с большой, то заменяет букву на большую;
* возраст должен быть числом от 18 до 60;
* ID - целое число, дополненное до 8 знаков незначащими нулями, ID должен быть уникальным.

Дополнительно: написать функцию, которая будет выводить полученный словарь в виде таблицы.

```python
def dict_output(dictionary): #вывод полученного словаря в виде таблицы
    if not dictionary: #если словарь не заполнялся - информирую, что он пустой
            print('Diary is empty')
    else:
        for id in dictionary.keys():
            name, lname, age = dictionary[id]
            print(f'| {id:10} | {name:10} | {lname:15} | {age} |')

diary = {}

while True:
    inp_str = input('Type <Name Last_name Age ID>, or press Enter to stop the loop: ')
    if inp_str: #для прерывания цикла
        allinfo = inp_str.split() #преобразую инпут в список
        if len(allinfo) == 4: #если в списке 4 элеемнта - дроблю его на строки 
            name, lname, age, user_id = allinfo
            try: #чек формата ИД - должен быть числом
                int_id = int(user_id)
                if int_id < 0:
                    print('Warning! ID must be a positive integer, try again')
                    continue
            except ValueError:
                print('ID must be int number, try again')
                continue
            
            if len(user_id) > 8: #чек входимости ИД в диапазон: если < 8 символов, дополняю ИД незначащими нулями
                print('Warning! ID must be <= 8, try again')
                continue
            full_id = user_id.zfill(8)

            if full_id in diary.keys(): #чек уникальности ИД
                print('Warning! ID must be unique, try again')
                continue

            if not (name.isalpha() and lname.isalpha()): #чек, что все символы имени и фамилии - буквы. Если да - привожу к верхнему регистру
                print('Warning! Name & Last name must be only alphas, try again')
                continue
            name, lname = name.capitalize(), lname.capitalize()
            
            if not age.isdigit(): #чек, что возраст - число
                print('Warning! Age must be a number')
                continue
            int_age = int(age)

            if not 18 <= int_age <= 60: #чек входимости возраста в диапазон
                print('Warning! Age must be between 18 and 60, try again')
                continue

            diary[full_id] = (name, lname, int_age) #если все проверки пройдены - данные записываются в словарь
                                       
        else:
            print('Warning! Wrong input, try again')
    else:
        dict_output(diary) #вывод словаря
        break
```
